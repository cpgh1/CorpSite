```python
# Tensor datatypes
# dtype is what datatyppe is the tensor (e.g. float32 or float16)
# device should match if you are running tensor code on one device and then attempt to run it on a different device
# e.g. running on "cpu" then attempting to run on gpu i.e. "cuda"
# requires_grad keeps tracking of gradients when set to true
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None,
                               device=None, # defaults to None, which uses the default tensor type of "cpu"
                               requires_grad=False) # if True, operations performed on the tensor are recorded
print(float_32_tensor)
print(float_32_tensor.dtype)

float_16_tensor = float_32_tensor.type(torch.float16)
print(float_16_tensor)

print(float_16_tensor * float_32_tensor)
```

```python
# Matrix Multiplication (Dot Product)
tensor = torch.tensor([1, 2, 3])

# Element wise multiplication
print(tensor, "*", tensor)
print("Equals:", tensor * tensor)

# Matrix multiplcation
torch.matmul(tensor, tensor) # 1*1 + 2*2 + 3*3 = 14

tensor @ tensor # You can also use @ from Python to do matrix multiplication
                # use matmul as that has been optimized -> see below

```

```python
# Shapes need to be in the right way
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]], dtype=torch.float32)

print(f"Original tensor_A shape = {tensor_A.shape} ")
print(f"Original tensor_B shape = {tensor_B.shape} ")
print("\n")

# torch.matmul(tensor_A, tensor_B) # (this will error)
#can also use torch.mm as it is an alias for torch.matmul

# View tensor_A and tensor_B.T (transposed tensor.B)
print(f"Still Original tensor_A = {tensor_A} ")
print(f"Transposed tensor_B i.e. tensor_B.T = {tensor_B.T} ")
print("\n")
print(f"tensor_A's shape = {tensor_A.shape} ")
print(f"Transposed tensor_b's shape = {tensor_B.T.shape} ")

# Now that the shapes satisfy the matrix multiplcation rules you can use matmul (or mm)
print("\n")
print(f"Matrix Multiplcation of tensor_A and transposed tensor_B = {torch.matmul(tensor_A,tensor_B.T)} ")
print("-------")
print(f"Shape of the output tensor = {torch.matmul(tensor_A,tensor_B.T).shape} ")
print(f"\n")
print(f"If you transpose tensor_A instead of tensor_B the the results are as follows")
print("\n")
print(f"Matrix Multiplcation of transposed tensor_A and original tensor_B = {torch.matmul(tensor_A.T,tensor_B)} ")
print("-------")
print(f"Shape of the output tensor = {torch.matmul(tensor_A.T,tensor_B).shape} ")
```

```python
# Tensor Aggregation (min,max, sum etc.)

x = torch.arange(0,100,10)
print(x)

# Find the min (can use torch.min or tensor.min() )
print(x.min())

# Find the max (can use torch.max or tensor.max() )
print(x.max())

# Find the mean
# print(x.mean()) <-- will thrown an error as input dtype is Long and mean must have
# either floating point or complex dtype
# So we need to convert to float32 first

print(x.type(torch.float32).mean())

# Find the sum
print(x.sum())

# positional min and mx using torch.argmin and torch.argmax
# The index value at which the min and max occur
print(f"\n")
print(x.argmin())
print(x.argmax())

print(f"\n")
# We can use the returned index to find the element at that position
x[9] # Gives the max
```

```python
# Reshaping, stacking, squeezing and unsqueezing tensors
# All of these are designed to change the shape or the dimension of the tensor in some way

import torch # Just allows us to run from this point rather than everything previous
x = torch.arange(1., 10.)
print(x , x.shape)

# Add an extra dimension
x_reshaped = x.reshape(1, 9)  # Dimensions have to be compatible with oriignal dimensions
print(x_reshaped) # Note the additional pair of [] in the reshaped tensor

x_reshaped = x.reshape(9, 1)
print(x_reshaped) # Note the [] around each element
# Difference is that (1,9) gave us one dimension with 9 elements
# while (9, 1) gave us 9 dimensions with 1 element each

# Change the view
z = x.view(1,9)
print(z, z.shape)

# Changing z changes x (becaue a view shares the same memeory as the oriignal tensor)
# z[:, 0] = 5 sets both the view z and tesnor x to be the same.
# In the above case the 0th element has its value changed to 5

#Stack tensors on top of each other
x_stacked = torch.stack([x,x,x,x], dim = 0)
print(x_stacked)
# Changing the dim changes the way the tesnors get stacked

# torch.squeeze() removes all single dimensions from a target tensor
x_squeezed = x_reshaped.squeeze()
print(x_squeezed)

# Note the double square brackets bfore squeeze [[..]] and single after squeeze
# Essentially a tensor with shape ([1,9]) has now become just ([9])
# i.e. The single dimension has been removed

# torch.unsqueeze() adds a single dimension to a target tensor at a specific dim (dimension)
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(x_unsqueezed.shape)

# torch.permute Rearrange the dimensions of a target tensor in a specified order
x_original = torch.rand(size=(224,224,3)) # [height, width, colour_channels ]

# Permute the original tensor to rearrange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1)
print(x_permuted.shape)


x_original[0,0,0] = 2255
print(x_original[0,0,0,], x_permuted[0,0,0])
# The permute shares the memory with the original so a change to the original is shared with
# with the permuted tensor

```

```python
# Indexing (selecting data from tensors)

x = torch.arange(1, 10).reshape(1, 3, 3)
print(x)
print(x.shape)

print(x[0])
print(x[0][0])
print(x[0][0][0])
print(x[0][2][2])

# You can also use ":" to select 'all' of a target dimension
print(x[:, 0])
print(x[:, :, 1])

print(x[:, 1, 1])

print(x[0, 0, :])

# Index on x to return 9
# Index on x to return 3,6,9
```
