```python
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
torch.__version__
```

```python
# scalar
scalar = torch.tensor(7)
print(scalar)
print(scalar.ndim)
print(scalar.item())
```

```python
# vector
vector = torch.tensor([7, 7])
print(vector)
print(vector.ndim)
print(vector.shape)
```

```python
# MATRIX
MATRIX = torch.tensor([[7, 8],
                       [9, 10]])
print(MATRIX)
print(MATRIX.ndim)
print(MATRIX[0])
print(MATRIX[1])
print(MATRIX[0][0])
print(MATRIX.shape)
```

```python
# TENSOR
# TENSOR shape gives 1,3,3 because the outer 1 is the 3x3 tensor itself
# We then have the 3 rows and 3 columns so 1,3,3
TENSOR = torch.tensor([[[1,2,3],
                        [3,6,9],
                        [2,5,4]]])
print(TENSOR)
print(TENSOR.ndim)
print(TENSOR.shape)
print(TENSOR[0])
```
